```python
even = open("even.txt","w+")
#even = open("even.txt","a") used to append the file instead of overwritting
odd = open("odd.txt","w+")
for i in range(1,101):
    if(i&1)==0:
        even.write(str(i)+" ")
    else:
        odd.write(str(i)+" ")
even.seek(0)#set the pointer to the first of the file
odd.seek(0)
print(even.read())
print(odd.read())
even.close()
odd.close()
```

```python
# imoporting data from the internet
import  urllib.request as web
import json
data = web.urlopen("https://gist.githubusercontent.com/prayagKhanal/8cdd00d762c48b84a911eca2e2eb3449/raw/5c5d62797752116799aacaeeef08ea2d613569e9/cakes.json")
gdata = data.read()#byte format of data
print(type(gdata))
s=gdata.decode()#converting byte data into the string datatype// byte to string
print(s)#string format of data
jdata = json.loads(s)
```
