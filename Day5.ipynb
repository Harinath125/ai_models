```python
class Demo:
    def __init__(self,a):
        self.num=a
    def  __add__(self,other):#magic method predefined by the python interpreter
        return self.num+other.num
    

a1=Demo(10)
a2=Demo(30)

a1+a2
```

```python
#check all the magic methods on w3 resources
class Complx:
    def __init__(self,r,i):
        self.real =r
        self.img =i
    def __add__(self,other):
        return self.real+other.real,self.img+other.img

c1=Complx(3,9)
c2=Complx(2,5)

c1+c2
```

```python
class Person:
    def __init__(self, name, designation="employee"):
        self.name = name
        self.designation = designation

    def learn(self):
        print(f"{self.name} is learning.")

    def walk(self):
        print(f"{self.name} is walking.")

    def eat(self):
        print(f"{self.name} is eating.")
class Programmer(Person):
    def code(self):
        print(f"{self.name} is coding.")
class Dancer(Person):
    def dance(self):
        print(f"{self.name} is dancing.")
class Singer(Person):
    def sing(self):
        print(f"{self.name} is singing.")
```

```python
p1 = Programmer("harinath")
p1.eat()
```

```python
class Electronic:
    def __init__(self):
        print("Electronic device")
class Phone(Electronic):
    def __init__(self):
        super().__init__()
        print("Phone")
    def make_call(self):
        print("Making a call...")
class Sound(Electronic):
    def __init__(self):
        super().__init__()
        print("Sound device")
    def play_sound(self):
        print("Playing sound...")
class Mobile(Phone):
    def __init__(self):
        super().__init__()
        print("Mobile phone")
    def browse_internet(self):
        print("Browsing internet...")
class Cord(Phone):
    def __init__(self):
        super().__init__()
        print("Corded phone")
    def dial_number(self):
        print("Dialing number on corded phone...")
class Earphones(Sound):
    def __init__(self):
        super().__init__()
        print("Earphones")
    def listen_music(self):
        print("Listening to music...")
class Stereo(Sound):
    def __init__(self):
        super().__init__()
        print("Stereo system")
    def play_music(self):
        print("Playing music on stereo...")
```

```python
c1=Stereo()
c1.play_sound()
```
