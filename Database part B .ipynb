```python
#!/usr/bin/python

import sqlite3

# Connect to SQLite database (creates one if not exists)
conn = sqlite3.connect('test.db')
print("Opened database successfully")

# Create table REGISTER if it doesn't already exist
conn.execute('''CREATE TABLE IF NOT EXISTS REGISTER
       (RID INTEGER PRIMARY KEY AUTOINCREMENT,
       UNAME TEXT NOT NULL,
       PWD TEXT NOT NULL,
       EMAIL TEXT,
       PHONE TEXT,
       AGE INT);''')
print("Table created successfully")

# Optional: Insert sample data (run once)
# Uncomment below if you want sample users in your database
'''
conn.execute("INSERT INTO REGISTER (UNAME, PWD, EMAIL, PHONE, AGE) VALUES ('harinath', '1234', 'harinath@gmail.com', '9876543210', 22)")
conn.execute("INSERT INTO REGISTER (UNAME, PWD, EMAIL, PHONE, AGE) VALUES ('anil', 'abcd', 'anil@gmail.com', '9123456789', 23)")
conn.commit()
print("Sample records inserted successfully")
'''

# Ask user to enter username and password
uname = input("Enter username: ")
pwd = input("Enter password: ")

# Query to verify user
cursor = conn.execute("SELECT EMAIL, PHONE FROM REGISTER WHERE UNAME=? AND PWD=?", (uname, pwd))
row = cursor.fetchone()

if row:
    print("\nLogin successful!")
    print("Email:", row[0])
    print("Phone:", row[1])
else:
    print("\nInvalid user. Please check your username and password.")

conn.close()
```
