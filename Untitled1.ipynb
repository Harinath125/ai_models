```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score


X, y = make_classification(
    n_samples=200,
    n_features=2,
    n_redundant=0,
    n_clusters_per_class=1,
    n_classes=2,
    random_state=42
)

# Split into training and testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)


knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)


y_pred = knn.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"✅ Model Accuracy: {accuracy * 100:.2f}%")


h = 0.02  # step size in mesh
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, h), 
                     np.arange(y_min, y_max, h))

# Predict classifications for each grid point
Z = knn.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Plot decision boundary
plt.figure(figsize=(8,6))
plt.contourf(xx, yy, Z, cmap=plt.cm.Paired, alpha=0.3)
plt.scatter(X_train[:, 0], X_train[:, 1], c=y_train, cmap=plt.cm.Paired, label="Training Data", edgecolor='k')
plt.scatter(X_test[:, 0], X_test[:, 1], c=y_test, cmap=plt.cm.Paired, marker='x', label="Test Data")
plt.title("K-Nearest Neighbors (KNN) Decision Boundary")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```

```python
# Example points
point1 = [2, 3]
point2 = [5, 7]

# Manhattan distance formula: sum of absolute differences
def manhattan_distance(p1, p2):
    distance = sum(abs(a - b) for a, b in zip(p1, p2))
    return distance

# Compute distance
dist = manhattan_distance(point1, point2)
print(f"Manhattan distance between {point1} and {point2} is {dist}")
```

```python
import numpy as np
from collections import Counter

# Sample dataset
# Columns: [Height, Weight]
X_train = np.array([
    [167, 51],
    [182, 62],
    [176, 70],
    [173, 64],
    [172, 65],
    [174, 57],
    [170, 55],
    [169, 58]
])

y_train = ["Underweight", "Normal", "Normal", "Normal", "Normal", "Normal", "Underweight", "Normal"]

# New point to classify
X_test = np.array([170, 57])

# Manhattan distance function
def manhattan_distance(a, b):
    return np.sum(np.abs(a - b))

# KNN function
def knn_predict(X_train, y_train, X_test, k=3):
    distances = []
    for x, y in zip(X_train, y_train):
        dist = manhattan_distance(x, X_test)
        distances.append((dist, y))
    # Sort distances
    distances.sort(key=lambda x: x[0])
    # Take k nearest neighbors
    neighbors = distances[:k]
    # Count the classes
    classes = [label for _, label in neighbors]
    most_common = Counter(classes).most_common(1)
    return most_common[0][0]

# Predict
predicted_class = knn_predict(X_train, y_train, X_test, k=3)
print(f"The predicted class for {X_test} is: {predicted_class}")
```

```python
# model.py
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
import joblib

# Load sample dataset
iris = load_iris()
X, y = iris.data, iris.target

# Train model
clf = RandomForestClassifier()
clf.fit(X, y)

# Save model to disk
joblib.dump(clf, "iris_model.pkl")
```

```python
# app.py
from flask import Flask, request, jsonify
import joblib
import numpy as np

app = Flask(__name__)

# Load the trained model
model = joblib.load("iris_model.pkl")

@app.route('/predict', methods=['POST'])
def predict():
    data = request.get_json()  # Expecting JSON: {"features": [5.1, 3.5, 1.4, 0.2]}
    features = np.array(data['features']).reshape(1, -1)
    prediction = model.predict(features)
    return jsonify({"prediction": int(prediction[0])})

if __name__ == "__main__":
    app.run(debug=True)
```
