```python
# Install dependencies first if not installed
# !pip install pythreejs ipywidgets trimesh numpy

import ipywidgets as widgets
from IPython.display import display
import trimesh
from pythreejs import *

# -----------------------------
# 1. Ask user for STL files
# -----------------------------
import tkinter as tk
from tkinter import filedialog

root = tk.Tk()
root.withdraw()
files = filedialog.askopenfilenames(
    title="Select STL Files",
    filetypes=[("STL files", "*.stl")]
)
if not files:
    print("No file selected")
    exit()

# -----------------------------
# 2. Load meshes
# -----------------------------
meshes = []
for f in files:
    tri_mesh = trimesh.load(f)
    vertices = tri_mesh.vertices
    faces = tri_mesh.faces
    geometry = BufferGeometry(
        attributes={
            'position': BufferAttribute(array=vertices, normalized=False),
            'index': BufferAttribute(array=faces.flatten(), normalized=False)
        }
    )
    material = MeshStandardMaterial(color='cyan', side='DoubleSide')
    mesh_obj = Mesh(geometry=geometry, material=material)
    meshes.append(mesh_obj)

# -----------------------------
# 3. Setup scene
# -----------------------------
scene = Scene(children=meshes + [AmbientLight(color='white', intensity=0.5),
                                 DirectionalLight(color='white', position=[5,5,5], intensity=0.6)])
camera = PerspectiveCamera(position=[5, 5, 5], up=[0,1,0], lookAt=[0,0,0], fov=50)
controller = OrbitControls(controlling=camera)

renderer = Renderer(camera=camera,
                    scene=scene,
                    controls=[controller],
                    width=800,
                    height=600)

display(renderer)

# -----------------------------
# 4. Add simple sliders for rotation/scale
# -----------------------------
rotation_slider = widgets.FloatSlider(value=0, min=0, max=360, description='Rotate Y')
scale_slider = widgets.FloatSlider(value=1, min=0.1, max=5, step=0.1, description='Scale')

def update_mesh(change):
    angle = rotation_slider.value
    scale = scale_slider.value
    for mesh in meshes:
        mesh.rotation = [0, np.radians(angle), 0]
        mesh.scale = [scale, scale, scale]

rotation_slider.observe(update_mesh, names='value')
scale_slider.observe(update_mesh, names='value')

display(rotation_slider, scale_slider)
```
